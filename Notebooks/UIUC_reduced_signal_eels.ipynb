## test dropping out

```python
# TODO: loop through and generate log2 spaced noise percentage 
file.data[np.isnan(file.data)] = 0

all_count = file.data.sum(axis=2).reshape(-1,1)
cumsum = np.cumsum(file.data.astype('float64'),axis=2).reshape(-1,file.data.shape[2])

keep = .0003
keeping = [np.random.randint(0,count,size=int(keep*count)) for count in all_count]
print("keeping counts about:", int(keep*all_count[10]))
keep_bins = [np.searchsorted(cumsum[i], drop) for i, drop in enumerate(keeping)]

bins_counts = [np.unique(bin, return_counts=True) for bin in keep_bins]

new_data = np.zeros((file.data.shape[0]*file.data.shape[1],file.data.shape[2]))
for i, (b, c) in enumerate(bins_counts):
    new_data[i][b] = c

new_file = file.copy()
new_file.data = new_data.reshape(file.data.shape)
```

```python
from Gaussian_Sampler.models.pv_fitter import Fitter_AE, pseudovoigt_1D_fitters, pseudovoigt_1D_fitters_new
from autophyslearn.spectroscopic.nn import block_factory, Conv_Block, FC_Block
from autophyslearn.spectroscopic.nn import Multiscale1DFitter
from Gaussian_Sampler.data.custom_sampler import Gaussian_Sampler



num_fits = 4 # number of curves to sum up
num_params = 4 # number of parameters to fit
dset = training_data#.reshape(30,30,-1)

fitter = Fitter_AE(function=pseudovoigt_1D_fitters_new,
                   dset=dset,
                   num_params=num_params,
                   num_fits=num_fits,
                   input_channels = 1,
                   learning_rate=3e-5,
                   device='cuda:0',
                   encoder = Multiscale1DFitter,
                   encoder_params = {
                       "model_block_dict": { # factory wrapper for blocks
                            "hidden_x1": block_factory(Conv_Block)(output_channels_list=[8,4], 
                                                                    kernel_size_list=[5,3], 
                                                                    pool_list=[128], 
                                                                    max_pool=False),
                            "hidden_xfc": block_factory(FC_Block)(output_size_list=[64,32]),
                            "hidden_x2": block_factory(Conv_Block)(output_channels_list=[4,4,4], 
                                                                    kernel_size_list=[3,3,3], 
                                                                    pool_list=[8,4], 
                                                                    max_pool=True),
                            "hidden_embedding": block_factory(FC_Block)(output_size_list=[8*num_fits,num_params*num_fits], last=True),
                        },
                        # TEST: LIMITS,
                        "skip_connections": {'hidden_xfc': 'hidden_embedding'},
                        "function_kwargs": {'limits': [1,dset.shape[-1],dset.shape[-1]] }
                    },
                    checkpoint_folder='../../checkpoints/reduced_signal/',
                    sampler = Gaussian_Sampler,
                    sampler_params = {'dset': dset, 
                                      'batch_size': 50, 
                                      'gaussian_std': 5, 
                                      'orig_shape': new_file.data.shape, 
                                      'num_neighbors': 10, },
                )
```
